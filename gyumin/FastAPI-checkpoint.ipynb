# FastAPI 사용하기

```python
# 1. 라이브러리 가져오기
from fastapi import FastAPI
```

```python
# 2. FastAPI를 기반으로 한 app 생성
app = FastAPI()

# 3. 요청을 할 수 있는 url - 함수를 생성
@app.get("/temp")
def temp():
    return "success"
```

```python
# 동일출저 정책을 풀어줄 수 있는 설정을 잡아보자
from fastapi.middleware.cors import CORSMiddleware

```

```python
# app == controller와 유사한 역할
# app에다가 middleware를 끼우는 작업 진행 --> 정책을 좀 풀어주려고!
app.add_middleware(
    # 동일출처정책과 관련한 미들웨어
    CORSMiddleware,
    # 접근을 허용할 출처(==접근을 허용할 url)
    allow_origins = ['http://localhost:8090'],
    allow_methods = ["*"],
    allow_headers = ["*"],
    allow_credentials = True
)
```

```python
# 크롤링 코드를 작성해보자!
# 1. 필요한 라이브러리 로딩 (selenium 사용 예정)
from selenium import webdriver as wb
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By

import time
```
